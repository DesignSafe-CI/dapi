# Controlling dapi's Logging

dapi reports what it is doing through Python's standard `logging` module under the `dapi` logger namespace. One call controls the verbosity globally:

| Level | You see |
|---|---|
| `INFO` (default) | Concise milestones: staging, uploads, job submitted, monitoring started |
| `DEBUG` | Everything: path translations, per-file transfers, request construction |
| `WARNING` | Problems only |
| `QUIET` | Nothing |

Log lines go to **stdout**, so they render as normal cell output (no red boxes) and stay in order with your prints and progress bars.

````python
import dapi
from dapi import DSClient

ds = DSClient()  # at the default INFO level you see: Authentication successful.
````

## Default: INFO

Routine internals stay quiet. Translating a path, for example, logs nothing at INFO:

````python
uri = ds.files.to_uri("/MyData/analysis")
print(uri)  # only your own print appears
````

## DEBUG: the full trace

The same call now shows its internals, prefixed with the module that logged them:

````python
dapi.set_log_level("DEBUG")
uri = ds.files.to_uri("/MyData/analysis")
# dapi.files DEBUG: Translated '/MyData/analysis' to 'tapis://designsafe.storage.default/...'
````

## QUIET: silence dapi entirely

````python
dapi.set_log_level("QUIET")
uri = ds.files.to_uri("/MyData/analysis")
print("dapi said nothing; the call still worked:", uri)
````

````python
dapi.set_log_level("INFO")  # back to the default
````

## Setting the level before import

Use the environment variable when you want a session-wide default without touching code — for example in a batch script or before starting Jupyter:

```bash
export DAPI_LOG_LEVEL=WARNING
```

## Per-module control

dapi never configures the root logger, so standard `logging` configuration applies to its namespace — useful in applications that embed dapi:

````python
import logging

logging.getLogger("dapi").setLevel(logging.WARNING)  # quiet dapi overall...
logging.getLogger("dapi.files").setLevel(logging.DEBUG)  # ...but trace one module

dapi.set_log_level("INFO")  # reset for the rest of this session
````

## What always prints

Functions whose purpose *is* output print directly regardless of the log level, since you called them to see their report:

- `job.print_runtime_summary()`
- `ds.jobs.interpret_status(...)`
- `ds.apps.find(..., verbose=True)`

Full reference: [Logging and Verbosity](https://designsafe-ci.github.io/dapi/docs/logging) in the dapi docs.
